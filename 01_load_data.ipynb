## <span style="color:#fff;">Load Data</span>
<div style="height:3px;width:48px;background:#2563eb;border-radius:2px;margin:2px 0 10px;"></div>

Reads the three raw files straight from Drive.

```python
import numpy as np
import pandas as pd
import random
import os
import pickle

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

from google.colab import drive
drive.mount('/content/drive')

BASE_PATH = '/content/drive/MyDrive/Projects/DSL/News Classification/'
DATA_PATH = BASE_PATH + 'DB/'
ARTIFACT_DIR = BASE_PATH + 'artifacts/'
os.makedirs(ARTIFACT_DIR, exist_ok=True)

train      = pd.read_csv(DATA_PATH + 'development.csv')
evaluation = pd.read_csv(DATA_PATH + 'evaluation.csv')
submission = pd.read_csv(DATA_PATH + 'sample_submission.csv')

print(f"✓ Loaded development/evaluation/sample_submission — ARTIFACT_DIR='{ARTIFACT_DIR}'")
print(f"  train {train.shape}, evaluation {evaluation.shape}, submission {submission.shape}")
```

## <span style="color:#fff;">Save Artifacts</span>
<div style="height:3px;width:48px;background:#2563eb;border-radius:2px;margin:2px 0 10px;"></div>



```python
with open(ARTIFACT_DIR + '01_load_data.pkl', 'wb') as f:
    pickle.dump({
        'train': train,
        'evaluation': evaluation,
        'submission': submission,
        'BASE_PATH': BASE_PATH,
        'SEED': SEED,
    }, f)

print(f"✓ Saved load-data artifacts to {ARTIFACT_DIR}01_load_data.pkl")
```
